```python
# Imports
from pathlib import Path
from datetime import datetime
import adaptive_polish.gis_measurement as gm
import tifffile
import numpy as np
import matplotlib.pyplot as plt

# Set parameters
model_path = Path(r"C:\Users\dmv31621\OneDrive - The Rosalind Franklin Institute\2024\00_Adaptive-milling\adaptive_polish\src\adaptive_polish\dl_segmentation\2024-02-24_0013_gis_lamela_crack_pytorch_AUnet.ptchkp")
gm.init_model_with_path(model_path)

%matplotlib qt
```

```python
sem_path = r"C:\Users\dmv31621\OneDrive - The Rosalind Franklin Institute\2024\00_Adaptive-milling\test_images\demo_images\05_honest-ray_SEM.tif"
sem_image = tifffile.imread(sem_path)
prediction = gm.segment(sem_image)
clean_prediction = gm.clean_prediction(prediction, 1.3020833333333335e-08)

fig, axs = plt.subplots(1,3)
axs[0].set_title("SEM")
axs[0].imshow(sem_image, cmap="Greys_r")
axs[1].set_title("prediction")
axs[1].imshow(prediction)
axs[2].set_title("cleaned GIS")
axs[2].imshow(clean_prediction)
```

```python
mask_anything = prediction > 0
mask_anything_largest_only = gm.keep_only_largest_object(mask_anything)
mask_crack = prediction == 3
mask_crack_inside_largest_object = np.multiply(
        mask_crack, 
        mask_anything_largest_only
    )

fig, axs = plt.subplots(1,4)
axs[0].set_title("Mask of anything")
axs[0].imshow(mask_anything)
axs[1].set_title("Mask anything, largest")
axs[1].imshow(mask_anything_largest_only)
axs[2].set_title("Crack only")
axs[2].imshow(mask_crack)
axs[3].set_title("Crack inside largest")
axs[3].imshow(mask_crack_inside_largest_object)
```
